```python
# Outlier Detection
```

```python
Q1 = learning["Learning_Hours_Per_Week"].quantile(0.25)
Q3 = learning["Learning_Hours_Per_Week"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = learning[
    (learning["Learning_Hours_Per_Week"] < lower_bound) |
    (learning["Learning_Hours_Per_Week"] > upper_bound)
]

print("Number of outliers:", len(outliers))
outliers[["Profile_ID", "Learning_Hours_Per_Week"]]
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    learning["Learning_Hours_Per_Week"],
    bins=20,
    kde=True
)

plt.axvline(
    learning["Learning_Hours_Per_Week"].median(),
    linestyle="--",
    label="Median = 7.9 hours"
)

plt.axvline(
    upper_bound,
    linestyle="--",
    label=f"Outlier threshold = {upper_bound:.2f} hours"
)

plt.title("Learning Hours Per Week - Distribution")
plt.xlabel("Learning Hours Per Week")
plt.ylabel("Number of Profiles")
plt.legend()
plt.show()
```

```python
# Bivariate Analysis
# Learning Hours vs Career Readiness
```

```python
learning_readiness = pd.merge(
    learning,
    career_readiness[["Profile_ID", "Career_Readiness_Score"]],
    on="Profile_ID",
    how="inner"
)

print(learning_readiness.shape)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=learning_readiness,
    x="Learning_Hours_Per_Week",
    y="Career_Readiness_Score"
)

plt.title("Learning Hours vs Career Readiness")
plt.xlabel("Learning Hours Per Week")
plt.ylabel("Career Readiness Score")

plt.show()
```

```python
# Experience vs Career Readiness

experience_readiness = pd.merge(
    profile[["Profile_ID", "Experience_Years"]],
    career_readiness[["Profile_ID", "Career_Readiness_Score"]],
    on="Profile_ID",
    how="inner"
)

# Create experience groups
experience_readiness["Experience_Group"] = pd.cut(
    experience_readiness["Experience_Years"],
    bins=[-1, 1, 3, 5, 8],
    labels=[
        "0–1 Years",
        "1–3 Years",
        "3–5 Years",
        "5–8 Years"
    ]
)

# Calculate average readiness for each group
experience_avg = (
    experience_readiness
    .groupby("Experience_Group")["Career_Readiness_Score"]
    .mean()
)

print(experience_avg)
```

```python
plt.figure(figsize=(9, 6))

plt.bar(
    experience_avg.index.astype(str),
    experience_avg.values
)

plt.title("Average Career Readiness by Experience Group")
plt.xlabel("Experience Group")
plt.ylabel("Average Career Readiness Score")

plt.show()
```

```python
# Project Complexity vs Career Readiness
project_readiness = pd.merge(
    projects[["Profile_ID", "Project_Complexity"]],
    career_readiness[["Profile_ID", "Career_Readiness_Score"]],
    on="Profile_ID",
    how="inner"
)

project_avg = (
    project_readiness
    .groupby("Project_Complexity")["Career_Readiness_Score"]
    .mean()
    .sort_values(ascending=False)
)

print(project_avg)
```

```python
plt.figure(figsize=(9, 6))

plt.bar(
    project_avg.index,
    project_avg.values
)

plt.title("Average Career Readiness by Project Complexity")
plt.xlabel("Project Complexity")
plt.ylabel("Average Career Readiness Score")

plt.show()
```

```python
# Technical Skill Level vs Career Readiness
skill_avg = (
    skills
    .groupby("Profile_ID")["Skill_Score"]
    .mean()
    .reset_index(name="Average_Skill_Score")
)

skill_readiness = pd.merge(
    skill_avg,
    career_readiness[["Profile_ID", "Career_Readiness_Score"]],
    on="Profile_ID",
    how="inner"
)

print(skill_readiness.head())
print(skill_readiness.shape)
```

```python
# Average Skill Score vs Career Readiness
skill_readiness["Skill_Level_Group"] = pd.cut(
    skill_readiness["Average_Skill_Score"],
    bins=[0, 40, 60, 80, 100],
    labels=[
        "0–40",
        "41–60",
        "61–80",
        "81–100"
    ]
)

skill_group_avg = (
    skill_readiness
    .groupby("Skill_Level_Group", observed=False)["Career_Readiness_Score"]
    .mean()
)

print(skill_group_avg)
```

```python
plt.figure(figsize=(9, 6))

plt.bar(
    skill_group_avg.index.astype(str),
    skill_group_avg.values
)

plt.title("Average Career Readiness by Technical Skill Level")
plt.xlabel("Average Technical Skill Score")
plt.ylabel("Average Career Readiness Score")

plt.show()
```
